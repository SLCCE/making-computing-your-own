# ⚠️Danger⚠️
While loops run until their condition is False. What happens if the condition is always True? The while loop will never stop and the program will never move forward. 

## Example
Here, "Done" will never be printed because the program will never get out of the while loop. \
To stop the code block from running, hit the small square that appears in place of the play button. \
Command to stop code in general: Ctrl + C

```python
while True:
    x = 2
print("Done")
```

To avoid our code not progressing and running forever, we just have to be careful and set a break condition.

```python
x = 1
while True:
    x += 1
    if x == 1000:
        break
print("Done")
print(x)
```
