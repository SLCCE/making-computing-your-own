# Maze Game!

Time to use everything you've learned -- especially loops -- to escape a few mazes.

Each maze is a 10x10 grid. You start in the top-left corner and need to reach the exit in the bottom-right corner.

You get four functions to move around:

* `left()`
* `right()`
* `up()`
* `down()`

If you walk into a wall, nothing happens -- you just stay where you are, so it's safe to experiment.

The maze itself uses symbols you'll see a lot in programming:

* `#` is a wall
* `.` is open space
* `o` is you
* `E` is the exit

Run the setup cell below once before trying any maze.

````python
# Run this once -- it sets up the maze engine used by every maze below.

def start_maze(grid):
    global MAZE, player_x, player_y
    MAZE = [list(row) for row in grid]
    player_x = 0
    player_y = 0

def _try_move(dx, dy):
    global player_x, player_y
    new_x = player_x + dx
    new_y = player_y + dy
    # Walking into a wall (or off the edge) just does nothing.
    if 0 <= new_x < 10 and 0 <= new_y < 10 and MAZE[new_y][new_x] != "#":
        player_x = new_x
        player_y = new_y

def left():
    _try_move(-1, 0)

def right():
    _try_move(1, 0)

def up():
    _try_move(0, -1)

def down():
    _try_move(0, 1)

def printResult():
    for row_index, row in enumerate(MAZE):
        line = ""
        for col_index, cell in enumerate(row):
            if col_index == player_x and row_index == player_y:
                line += "o"
            else:
                line += cell
        print(line)
````

## Maze 0 -- Warm-up

Same shape as Maze 1, just flipped: straight down first, then straight right.

```
.#########
.#########
.#########
.#########
.#########
.#########
.#########
.#########
.#########
.........E
```

````python
start_maze([
    ".#########",
    ".#########",
    ".#########",
    ".#########",
    ".#########",
    ".#########",
    ".#########",
    ".#########",
    ".#########",
    ".........E",
])

# The "one call at a time" way -- it works, but it's a lot of typing.
down()
down()
down()
down()
down()
down()
down()
down()
down()
right()
right()
right()
right()
right()
right()
right()
right()
right()

printResult()
````

That worked, but it took 18 lines. Here's the same maze solved with a loop instead -- same result, in 2-4 lines. (`start_maze(...)` resets the player back to the top-left before trying again.)

````python
start_maze([
    ".#########",
    ".#########",
    ".#########",
    ".#########",
    ".#########",
    ".#########",
    ".#########",
    ".#########",
    ".#########",
    ".........E",
])

for i in range(9):
    down()
for i in range(9):
    right()

printResult()
````
